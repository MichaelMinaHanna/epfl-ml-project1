```python
import os

import matplotlib.pyplot as plt
import numpy as np

ridge_lambdas = np.logspace(-8, 0, 9)
ridge_f1s = [0.4283, 0.4272, 0.4279, 0.4281, 0.4277, 0.4269, 0.4266, 0.4256, 0.4205]

reg_logistic_lambdas = [1e-4, 1e-3, 1e-2]
reg_logistic_f1s = [0.4377, 0.4375, 0.4355]

logistic_f1 = 0.4378

fig, ax = plt.subplots(figsize=(3.4, 2.4))
ax.semilogx(ridge_lambdas, ridge_f1s, marker="o", label="ridge")
ax.semilogx(reg_logistic_lambdas, reg_logistic_f1s, marker="s",
            label="reg. logistic")
ax.axhline(logistic_f1, color="gray", linestyle="--",
           label=r"logistic (no $\lambda$)")
ax.set_xlabel(r"$\lambda$")
ax.set_ylabel("CV F1")
ax.grid(True, alpha=0.3)
ax.legend(fontsize=7)
fig.tight_layout()

os.makedirs("figures", exist_ok=True)
fig.savefig("figures/f1_vs_lambda.pdf")
print("Saved figures/f1_vs_lambda.pdf")
```
